```python
# step 1 impoer require librerie 

import pandas as pd 
import numpy as np 
```

```python
# step 2 load the dataset 

df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
#step 5 cheack the duplicate Record 
df.duplicated().sum()
```

```python
#remove duplicate 

df.drop_duplicates(inplace = True)
```

```python
df.duplicated().sum()
```

```python
#step 6 Replace invalid values with NAN
#replace comon invalid values with nan
df.replace(['N/A','NULL',''],np.nan,inplace = True)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df[df.apply(lambda row: row.astype(str).str.strip().eq('').any(), axis=1)]
```

```python
df.replace(r'^\s*$', np.nan, regex=True, inplace=True)
```

```python
#step 7 remove extra space 
df = df.apply(lambda x : x.str.strip() if x.dtype =='str' else x )
```

```python
#step 8 standardize Text Format 
# Convert text into Proper Case 

df['Customer_Name'] = df['Customer_Name'].str.title() 
df['City'] = df['City'].str.title() 
df['State'] = df['State'].str.title() 
```

```python
#step 9 valid emailid 
# keep only velid
df = df[df['Email'].str.contains('@',na=False)]
```

```python
#step 10 conver datecolumn 
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce') #nat
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce') #nat
```

```python
#step 11 convert numeric Columns 
cols = ['Qty','Unit_Price','Discount_Percent']

for c in cols :
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
#step 12 remove invalid Quantity 
#remove records having nagative or zero quantity 

df = df[df['Qty'] > 0]
```

```python
# step 13 full the mising values 

# 1️⃣ Fill missing discounts with 0
df['Discount'] = df['Discount'].fillna(0)

# 2️⃣ Fill missing phone numbers with 'Unknown'
df['Phone'] = df['Phone'].astype('object').fillna('Unknown')

# 3️⃣ Fill missing delivery dates with the corresponding order date
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])



```

```python
df.head()
```

```python
# step 14 : Feature Engineering 
# #total sales 

df['Sales']= df['Qty'] * df['Unit_Price']
```

```python
df.head()
```

```python
#net ammount

# Ensure Sales and Discount are numeric
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')


# Now safely calculate Net_Amount
df['Net_Amount'] = df['Sales'] - (df['Sales'] * df['Discount'] / 100)
```

```python
df.head()
```

```python
# profit 20 % 

df['Profit'] = df['Net_Amount'] * 0.20
```

```python
#order month 

# Convert Order_Date to datetime
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')

# Now you can use .dt.month_name()
df['Month'] = df['Order_Date'].dt.month_name()
```

```python
df
```

```python
#15.order year 

df['Year'] = df[ 'Order_Date'].dt.year
```

```python
#weekday

df['Weekday'] = df[ 'Order_Date'].dt.day_name()
```

```python
#step 16 Export Clean Data 
df.to_csv('Clean_Ecomarce.csv' , index=False)
```
